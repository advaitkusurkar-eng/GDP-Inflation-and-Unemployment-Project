```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("CSV_1.csv")
```

```python
df1
```

```python
df1 = df1.fillna(0)
df1 = df1.drop_duplicates()
df1 = df1.drop(columns = ['Indicator Name', 'Indicator Code','2025'])
```

```python
df1
```

```python
df2 = pd.read_csv("CSV_2.csv")
```

```python
df2
```

```python
df2 = df2.drop_duplicates()
```

```python
df2
```

```python
df2 = df2.drop(columns = ['Unnamed: 5'])
```

```python
df2 = df2.fillna("NA")
```

```python
df2
```

```python
df2.set_index('TableName', inplace=True)
```

```python
df2
```

```python
df3 = pd.merge(df1, df2, on = 'Country Code')
```

```python
df3
```

```python
df3.drop_duplicates(inplace=True)
```

```python
df3
```

```python
df3.set_index('Country Name', inplace=True)
```

```python
df3
```

```python
drop_columns = df3.loc[:,"1960":"1989"]
df3.drop(columns = drop_columns, inplace=True)
```

```python
df3
```

```python
df3[df3.select_dtypes(include='float64').columns] = df3.select_dtypes(include='float64').astype(int)
```

```python
df3
```

```python
df3.drop(columns = ['SpecialNotes', 'Region'], inplace=True)
```

```python
df3
```

```python
df3.to_csv('GDP.csv', index=True)
```

```python
df4 = pd.read_csv('CSV_4.csv')
```

```python
df4
```

```python
df4 = df4.fillna(0)
df4.drop(columns = ['Indicator Name', 'Indicator Code', '2025'], inplace=True)
df4.drop(columns = [str(year) for year in range(1960, 1990)], inplace=True)
```

```python
df4
```

```python
df4.drop_duplicates(inplace=True)
df4.set_index('Country Name', inplace=True)
df4.to_csv('Inflation.csv', index=True)
```
